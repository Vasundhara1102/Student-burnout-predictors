```python
import pandas as pd, seaborn as sns, matplotlib.pyplot as plt
df = pd.read_csv("StressLevelDataset.csv")

print(df.shape); print(df.isna().sum().sum()); print(df.duplicated().sum())
df["stress_level"].value_counts().plot(kind="bar", title="Stress level distribution")
plt.show()

plt.figure(figsize=(12,9))
sns.heatmap(df.corr(), cmap="coolwarm", center=0)
plt.show()

# Strongest relationships with stress
print(df.corr()["stress_level"].sort_values(ascending=False))
```

```python
sns.boxplot(x="stress_level", y="sleep_quality", data=df)
plt.title("Sleep quality vs stress level")
plt.show()
```
